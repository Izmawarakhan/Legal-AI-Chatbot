```python
import os
import pdfplumber
import json
```

```python
folders = ["01_Raw_PDFs", "02_Extracted_Text", "03_Final_Dataset"]
for folder in folders:
    if not os.path.exists(folder):
        os.makedirs(folder)
        print(f"Folder created {folder}")
```

```python
# 2. Extraction Code
def start_extraction():
    raw_path = "01_Raw_PDFs/"
    output_path = "02_Extracted_Text/"
    all_data = []

    # Check karein agar folder khali toh nahi
    files = [f for f in os.listdir(raw_path) if f.endswith('.pdf')]
    if not files:
        print("Error: '01_Raw_PDFs' folder is empty, please upload the pdfs")
        return

    for filename in files:
        print(f"Processing: {filename}...")
        with pdfplumber.open(os.path.join(raw_path, filename)) as pdf:
            text = ""
            for page in pdf.pages:
                text += page.extract_text() + "\n"
            
            # Save as Text File
            txt_name = filename.replace(".pdf", ".txt")
            with open(os.path.join(output_path, txt_name), "w", encoding="utf-8") as f:
                f.write(text)
            
            # Collect for JSON/MongoDB
            all_data.append({
                "law_name": filename,
                "content": text
            })

    # Save Final JSON
    with open("03_Final_Dataset/legal_data.json", "w", encoding="utf-8") as j:
        json.dump(all_data, j, indent=4)
    print("Congrats! JSON file '03_Final_Dataset' is created.")
```
